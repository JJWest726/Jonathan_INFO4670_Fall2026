### Q1 · One quantitative variable
**How many hours per week do Northgate students work?** Describe what's typical, how spread out it is, and whether anyone stands apart.
→ *histogram* (add a *box plot* if it helps). Column: `work_hours_per_week` (from `students`).

```python

fig, ax = plt.subplots()
sns.histplot(data=students, x="work_hours_per_week", color=NAVY, ax=ax)
ax.set_title("Hours worked by Students")
ax.set_xlabel("Work Hours")
ax.set_ylabel("Number of Students")
plt.show()


```

**STAR story — Q1** Each student seems to work any where from between 0 to 40 hours with a majority working in the ranges between 10 and 20 hours per week. One student stands out as working less a negative value of hours.

- **Scope —** This chart shows the number of hours a certain number of students work per week
- **Track —** How many hours students work per week
- **Articulate —** Each student seems to work any where from between 0 to 40 hours with a majority working in the ranges between 10 and 20 hours per week.
- **Respond —** There seems to be an error in the data as one student has -10 hours worked per week

### Q2 · One categorical variable
**How does enrollment compare across majors?**
→ *sorted bar chart*. Column: `major` (from `students`).

```python
order = students["major"].value_counts().index
fig, ax = plt.subplots()
sns.countplot(data=students, y="major", order=order, color=NAVY, ax=ax)   # which chart? which column(s)?
ax.set_title("Number of students by major")
ax.set_xlabel("Students")
ax.set_ylabel("Major")
plt.show()

```

**STAR story — Q2** Enrollemnt is quite close across majors Except for Cyber Security which has an extreme majority in comparision to the others

- **Scope —** This chart shows the number of students sorted by their major
- **Track —** How does enrollment look across majors
- **Articulate —** A majority of this chart is shown to be in cyber security major with the other majors being quite cloly tied together
- **Respond —** There are duplicate catagories caused by different spellings of the same major

### Q3 · A quantity, split by category
**Does final GPA differ across majors?** Read the exceptions, not just the boxes.
→ *box plot per major*. Columns: `final_gpa` and `major` (from `students`).

```python
fig, ax = plt.subplots()
sns.boxplot(data=students, x="final_gpa", y="major", color=NAVY, ax=ax)
ax.set_title("Final gpa by major")
ax.set_xlabel("Final GPA")
ax.set_ylabel("Major")
plt.show()


```

**STAR story — Q3** Grade medians and inner Q values do not differ too much from each other with cyber security having the highest median value though that may balance out when the duplicate values are removed. Information science has the largest range of Q values. Information Science also does not have any visiable outliers

- **Scope —** This chart shows final GPA based on majors
- **Track —** Do grades differ based on major
- **Articulate —** The typical values range between 1.9 and 2.8
- **Respond —** Duplicate data due to typos

### Q4 · Two quantitative variables
**Is there a relationship between how far students commute and their final GPA?** Remember: *association is not causation.*
→ *scatterplot with a trend line*. Columns: `commute_miles` and `final_gpa` (from `students`).

```python
fig, ax = plt.subplots()
sns.regplot(data=students, x="commute_miles", y="final_gpa",
            scatter_kws={"s": 12, "color": NAVY, "alpha": 0.4},
            line_kws={"color": RUST}, ax=ax)
ax.set_title("Final GPA vs. Commute Miles")
ax.set_xlabel("Commute miles")
ax.set_ylabel("Final GPA")
plt.show()
```

**STAR story — Q4** There appears to be some sort of relationship with an increase in commute miles and a decrease in final GPA
- **Scope —** This chart shows Final GPA compared to commuter miles with an included trend line
- **Track —** What if any relationship is there with Commute miles and Final GPA
- **Articulate —** A majority of students live within 10-20 miles of campus and the GPA clusters around 2.7
- **Respond —** There are several points that show GPA as 0.0

### Q5 · Over time
**How does student engagement change across the semester?**
→ *line chart* of average weekly activity. Use the `weekly` table: `week` and `minutes_active` (you'll need to average by week first — see the Guided notebook).

```python
weekly_avg = weekly.groupby("week")["minutes_active"].mean().reset_index()
fig, ax = plt.subplots()
sns.lineplot(data=weekly_avg, x="week", y="minutes_active",
             color=NAVY, marker="o", ax=ax)
ax.set_title("Average active minutes per student, by week")
ax.set_xlabel("Week of semester")
ax.set_ylabel("Avg. minutes active")
plt.show()
```

**STAR story — Q5** Student Active minuets stays at a constant level through out the Semester but drops off completly at week 9 before returning to a standard level

- **Scope —** Active student hours by week
- **Track —** How does student engagement change across the semester
- **Articulate —**  Student Active minuets stays at a constant level through out the Semester but drops off completly at week 9 before returning to a standard level
- **Respond —** There is a week 16

Pose **one question of your own** about the Northgate students, choose the chart that answers it, build it, label it, and write its STAR story. Going beyond Part A is rewarded — a per-student line chart, a violin plot, or a chart of a messier column are all fair game.

**Your question:** What additional insight can be gained by comparing individual student hours across the weeks instead of averages



```python
sample_ids = weekly["student_id"].drop_duplicates().sample(40, random_state=1)
sample = weekly[weekly["student_id"].isin(sample_ids)]
fig, ax = plt.subplots()
sns.lineplot(data=sample, x="week", y="minutes_active",
             units="student_id", estimator=None,
             color="grey", alpha=0.4, ax=ax)
ax.set_title("Weekly active minutes — 40 individual students")
ax.set_xlabel("Week of semester")
ax.set_ylabel("Minutes active")
plt.show()
```

**STAR story — Part B** Individual student active hours are alot more varied than the average per week

- **Scope —** 40 individual student's active minutes order by week
- **Track —** What insight can be gained from individual student active time instead of the average
- **Articulate —** The chart is very sparadic but has a large dip around the week 9 timeframe
- **Respond —** I would most certainly next look at the relationship with this graph and final GPA
